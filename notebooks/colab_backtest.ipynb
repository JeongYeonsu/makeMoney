## 5) 누적수익 곡선

```
import glob, matplotlib.pyplot as plt
for f in glob.glob('results/*_train/daily.csv'):
    d = pd.read_csv(f, parse_dates=['date'])
    plt.plot(d['date'], (1 + d['ret']).cumprod(), label=f.split('/')[1])
plt.legend(); plt.title('누적 수익'); plt.show()
```
